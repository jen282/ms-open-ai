```python
import requests
import dotenv
import os
```

```python
# OPEN API KEY 불러오기
dotenv.load_dotenv()
OPEN_AI_KEY = os.getenv("OPEN_AI_KEY")
```

```python
# Endpoint: OpenAI 호출하기 위한 Endpoint URI, URL
# 이건 별도의 파일에 관리하기도 함
endpoint = "https://9ai043-openai.openai.azure.com/openai/deployments/9ai043-gpt-4o-mini/chat/completions?api-version=2025-01-01-preview"

# Method : Post

# Header: OpenAI API Key 포함된 헤더 정보
headers = {
    "Authorization": OPEN_AI_KEY,
    "Content-Type": "application/json"
}
# Body: Body의 메시지 정보
body = {
        "messages": [
            {
                "role": "system",
                "content": "긍정/부정을 판별해줘"
            },
            {
                "role": "user",
                "content": "오늘 날씨가 흐려서 기분이 다운됐어"
            },
            {
                "role": "assistant",
                "content": "부정"
            },
            {
                "role": "user",
                "content": "오늘 햇살이 너무 좋아서 기분이 좋아졌어"
            },
            {
                "role": "assistant",
                "content": "긍정"
            },
            {
                "role": "user",
                "content": "날씨가 흐려 "
            },
            {
                "role": "assistant",
                "content": "중립"
            },
            {
                "role": "user",
                "content": "아니 근데 이게 말이 돼?"
            },
            {
                "role": "assistant",
                "content": "부정"
            },
            {
                "role": "user",
                "content": "정말 대단하시네요^^"
            }
        ],
        "max_tokens": 4096,
        "temperature": 1,
        "top_p": 1,
        "model": "9ai043-gpt-4o-mini"
        }

# OpenAI API 호출
response = requests.post(endpoint, headers=headers, json=body)
response_json = response.json()
content = response_json['choices'][0]['message']['content']
content
# print(response)
```
