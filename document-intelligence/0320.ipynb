#### 3-2. GET 요청 보내기

```python
def request_document_intelligence(image_path):
    endpoint = AZURE_DOcUMENT_ENDPOINT
    headers = {
        "Ocp-Apim-Subscription-Key": AZURE_DOCUMENT_API_KEY,
        "Content-Type": "image/*"  # 이걸 안하면 400: 
     }

    # 전송을 위해 image_path -> binary data로 변환 필요
    with open(image_path, "rb") as image_file:    # rb: read binary / w:write / x:execute
        image_data = image_file.read()
        
    body = image_data
    # print(body) # 여기에 디버깅 걸어두고 디버깅해보기

    # POST 요청 보내기 
    response = requests.post(endpoint, headers=headers, data=body) # json, xml도 아닌 이미지 자체를 날림
    
    if response.status_code != 202:
        print(f"Error: {response.status_code} - {response.text}")
        return None
    
    print(f"status code: {response.status_code}")

    # POST 요청을 하면 resonese의 헤더로 주는 이미지 분석 결과의 주소를 가져온다 
    # 방법1)
    # result_url = response.headers['Operation-Location']  

    # 방법2) .get을 사용하면 키값이 없어도 에러가 안 남. 값이 없을때 리턴할 값도 설정할 수 있다
    result_url = response.headers.get('Operation-Location', False) #디폴트 리턴 값은 None
    print(f"operation-location: {result_url}")

    # 분석이 완료되는 시점. succeeded 또는 failed 될 때까지 반복해서 상태를 확인 => 3-2. GET 요청에서 구현
    # succeeded / failed가 되면 while문을 빠져나와 결과를 리턴한다.
    while True:
        result_response = requests.get(result_url, headers=headers) # 헤더의 콘텐츠 타입과 키값은 동일하므로 POST에서 정의한 headers 그대로 사용
        
        # 응답 예외 처리
        if not result_response.ok:
            print(f'Error: {response.status_code}-{response.text}')
        
        result_json = result_response.json()
        current_status = result_json['status']
        
        if current_status == "running":
            print(f"current status: {current_status}", flush=True)
            time.sleep(1) # 1초 후 다시 상태 확인
            continue
        else:
            break
        
    print(f"final status: {current_status}")
    
    if current_status != "succeeded":
        return None
    

    for k in ['status', 'createdDateTime','lastUpdatedDateTime','analyzeResult']:
        print(f'{k}: {result_json[k]}\n')

    return result_json
        

with gr.Blocks() as demo:

    # 클릭 이벤트 정의
    def click_send(image_path):
        print(image_path)
        request_document_intelligence(image_path)
        # image_path -> pil image로 변환하는 로직 
        return image_path, image_path

    gr.Markdown("## Doucment Intelligence")

    # 인풋, 아웃풋 이미지 컴포넌트 2개 좌우로 배치하기
    with gr.Row():
        input_image = gr.Image(label="Input Image", type="filepath")
        # output_image = gr.Image(label="Output Image", type="filepath")
        output_image = gr.Image(label="Output Image", interactive=False, type="pil")

    path_text = gr.Textbox(label="Image Path", interactive=False)
    # 전송 버튼 추가하기
    send_button = gr.Button("전송")
    send_button.click(click_send, inputs=[input_image], outputs=[output_image, path_text])

# demo.launch()

# 테스트용 함수
click_send("./통보서.jpg") #상태 체크를 확인하기 위해 용량이 큰 문서로 시도 
```

#### 4-2. bounding box 그리기

- 결과 데이터에서 필요한 정보를 추출해 이미지에 그린다
* 결과 데이터에서 텍스트, 위치 정보를 추출해 이미지에 텍스트를 그린다 

```python
def request_document_intelligence(image_path):
    endpoint = AZURE_DOcUMENT_ENDPOINT
    headers = {
        "Ocp-Apim-Subscription-Key": AZURE_DOCUMENT_API_KEY,
        "Content-Type": "image/*"  # 이걸 안하면 400: 
     }

    # 전송을 위해 image_path -> binary data로 변환 필요
    with open(image_path, "rb") as image_file:    # rb: read binary / w:write / x:execute
        image_data = image_file.read()
        
    body = image_data
    # print(body) # 여기에 디버깅 걸어두고 디버깅해보기

    # POST 요청 보내기 
    response = requests.post(endpoint, headers=headers, data=body) # json, xml도 아닌 이미지 자체를 날림
    
    if response.status_code != 202:
        print(f"Error: {response.status_code} - {response.text}")
        return None
    
    print(f"status code: {response.status_code}")

    # POST 요청을 하면 resonese의 헤더로 주는 이미지 분석 결과의 주소를 가져온다 
    # 방법1)
    # result_url = response.headers['Operation-Location']  

    # 방법2) .get을 사용하면 키값이 없어도 에러가 안 남. 값이 없을때 리턴할 값도 설정할 수 있다
    result_url = response.headers.get('Operation-Location', False) #디폴트 리턴 값은 None
    print(f"operation-location: {result_url}")

    # 분석이 완료되는 시점. succeeded 또는 failed 될 때까지 반복해서 상태를 확인 => 3-2. GET 요청에서 구현
    # succeeded / failed가 되면 while문을 빠져나와 결과를 리턴한다.
    while True:
        result_response = requests.get(result_url, headers=headers) # 헤더의 콘텐츠 타입과 키값은 동일하므로 POST에서 정의한 headers 그대로 사용
        
        # 응답 예외 처리
        if not result_response.ok:
            print(f'Error: {response.status_code}-{response.text}')
        
        result_json = result_response.json()
        current_status = result_json['status']
        
        if current_status == "running":
            print(f"current status: {current_status}", flush=True)
            time.sleep(1) # 1초 후 다시 상태 확인
            continue
        else:
            break
        
    print(f"final status: {current_status}")
    
    if current_status != "succeeded":
        return None
    

    for k in ['status', 'createdDateTime','lastUpdatedDateTime','analyzeResult']:
        print(f'{k}: {result_json[k]}\n')

    return result_json
        

def draw_image(image_path, result_data):
    # PIL 이미지 생성
    image = Image.open(image_path)

    # 그리기 위한 draw(붓) 생성하기
    draw = ImageDraw.Draw(image)

    # 결과 이미지가 PIL로 잘 불러와졌는지 확인하기 위해 테스트로 글씨 써보기 
    # draw.text((100,200), "Document Intelligence", fill="red")

    # 폴리곤 정보 뽑아내기 (디버깅하면서 어디에 담겨있는지 확인 후 코드 작성)
    analyzeResult = result_data['analyzeResult']
    block_lists = analyzeResult['paragraphs']

    for block in block_lists:

        content = block.get('content')
        polygon = block.get('boundingRegions')[0].get('polygon')
        # print(content, polygon) 

        # 폴리곤 정보를 이용해 bounding box 그리기
        polygon_list = list(zip(polygon[::2], polygon[1::2]))  # 폴리곤 좌표 추출
    
        # 폴리곤 색상 및 폰트 지정
        line_color = random_col()
        font = get_font(abs(polygon_list[1][0]-polygon_list[0][0]))

        draw.polygon(polygon_list, outline=line_color, width=3)
        draw.text(polygon_list[0], content, fill=line_color, font=font, width=1)
    
    return image
##==== 폴리곤 박스 랜덤 색상 지정 ====
def random_col():
    # 랜덤한 RGB 튜플 반환
    import random
    return (random.randint(0, 255), random.randint(0, 255), random.randint(0, 255))

##==== 한글설정 ====
def get_font(width):
    # OS별로 적절한 폰트 객체 반환 
    from PIL import ImageFont
    import platform

    font_size = max(8, int((width/900)* 20))
    try:
        if platform.system() == "Windows":
            return ImageFont.truetype("malgun.ttf", font_size)
        elif platform.system() == "Darwin": # MacOS
            return ImageFont.truetype("AppleGothic.ttf", font_size)
        else: # Linux
            return ImageFont.load_default(size=font_size)
    except (OSError, IOError):
        # 폰트파일이 없는 경우 기본 폰트 사용
        return ImageFont.load_default(size=font_size)

with gr.Blocks() as demo:

    # 클릭 이벤트 정의
    def click_send(image_path):
        result_data = request_document_intelligence(image_path)
        image = draw_image(image_path, result_data)
        # image_path -> pil image로 변환하는 로직 
        # return image, image_path 
        return image # 파일경로는 빼고 리턴

    gr.Markdown("## Doucment Intelligence")

    # 인풋, 아웃풋 이미지 컴포넌트 2개 좌우로 배치하기
    with gr.Row():
        input_image = gr.Image(label="Input Image", type="filepath")
        # output_image = gr.Image(label="Output Image", type="filepath")
        output_image = gr.Image(label="Output Image", interactive=False, type="pil")

    # path_text = gr.Textbox(label="Image Path", interactive=False) # 파일경로 삭제
    # 전송 버튼 추가하기
    send_button = gr.Button("전송")
    # send_button.click(click_send, inputs=[input_image], outputs=[output_image, path_text])
    send_button.click(click_send, inputs=[input_image], outputs=[output_image]) #파일 경로는 제외햇으므로 이것도 제외

demo.launch()

# 테스트용 함수
# click_send("./통보서.jpg") #상태 체크를 확인하기 위해 용량이 큰 문서로 시도 
```
